# Construcción dataset

Se busca construir un dataframe compuestos por frames de videos recolectados en la visita de campo a los galpones asigandos para el conteo de aves.


<br />


```python
import os
os.chdir(r"B:\11RossyDocumentos\18ProyectoPollos\12Avances\11_proyecto_pollos")

import cv2
from pathlib import Path
```

## Carga de imágenes

```python
list_videos = os.listdir("data/raw")
list_videos
```

```python
ruta_videos = Path("data/raw")
ruta_frames = Path("data/processed")

intervalo_segundos = 20
# intervalo_segundos = 30

resumen = list()

for i in list_videos:
    video_path = ruta_videos / i
    video_stem = video_path.stem

    frame_path = ruta_frames / video_stem
    frame_path.mkdir(parents=True, exist_ok=True)

    cap = cv2.VideoCapture(str(video_path))
    fps = cap.get(cv2.CAP_PROP_FPS)
    total_frames = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))

    duracion_segundos = total_frames / fps
    timestamps = range(0, int(duracion_segundos) + 1, intervalo_segundos)

    frames_guardados = 0

    for segundo in timestamps:
        # Ubica el video en el segundo indicado
        cap.set(cv2.CAP_PROP_POS_MSEC, segundo * 1000)

        exito, frame = cap.read()

        if exito:
            nombre_frame = f"{video_stem}_t{segundo:04d}s_{frames_guardados+1}.jpg"
            ruta_frame = frame_path / nombre_frame

            cv2.imwrite(str(ruta_frame), frame)
            frames_guardados += 1

    cap.release()

    resumen.append({
        "video": i,
        "duracion_minutos": round(duracion_segundos / 60, 2),
        "frames_guardados": frames_guardados
    })
```
